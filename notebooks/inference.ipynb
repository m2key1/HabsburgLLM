```python
from IPython.display import display, Markdown
import torch
from transformers import AutoTokenizer, Olmo3ForCausalLM
```

```python
model_path = "M2key1/HabsburgLLM"

device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = Olmo3ForCausalLM.from_pretrained(model_path, dtype=torch.bfloat16).to(device)
model.eval()
```

```python
prompt = """Wenn ich ein Flugzeug beschreiben muesste, dann """.replace("\n", " ")

inputs = tokenizer(prompt, return_tensors="pt").to(device)

output = model.generate(  # type: ignore[attr-defined]
    **inputs,
    max_new_tokens=128,
    do_sample=True,
    top_k=20,
    temperature=0.5,
    top_p=0.95,
    repetition_penalty=1.1,
    no_repeat_ngram_size=6
)

output = tokenizer.decode(output[0], skip_special_tokens=True)

display(Markdown(f"**prompt:** {prompt}...\n\n**output:** ...{output[len(prompt):]}..."))
```
